#### Plot time series

```python
import matplotlib.pyplot as plt
n.statistics.energy_balance(aggregate_time=False, comps=["Store"]).droplevel(0).iloc[
    :, :
].groupby("carrier").sum().T.plot.line()
# .loc[:, :, "el"].droplevel(0).iloc[
#     :, :4000
# ].groupby("carrier").sum().where(lambda x: np.abs(x) > 1).fillna(0).T.plot.area(
#     ax=ax, title="Energy Balance Timeseries"
# )
# ax.legend(bbox_to_anchor=(1, 0), loc="lower left", title=None, ncol=1)
```

```python
fig, ax = plt.subplots()
n.statistics.energy_balance(aggregate_time=False).loc[:, :, "el"].droplevel(0).iloc[
    :, :4000
].groupby("carrier").sum().where(lambda x: np.abs(x) > 1).fillna(0).T.plot.area(
    ax=ax, title="Energy Balance Timeseries"
)
ax.legend(bbox_to_anchor=(1, 0), loc="lower left", title=None, ncol=1)
```
