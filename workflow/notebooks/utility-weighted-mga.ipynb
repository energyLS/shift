```python
import pandas as pd
```

```python
# Define threshold to select instable regions
threshold = -0.5
```

```python
stability = pd.read_csv("../../resources/political_stability_clustered.csv", index_col=0)
```

```python
stability.head()
```

### Filter according to threshold and compute weighting

```python
instable = stability[stability.political_stability < threshold].copy()
```

### Use worst region as baseline (which is 1) and assign weighting to others accordingly

```python
instable["weight"] = instable.political_stability / instable.political_stability.min()
```

```python
instable.weight.sort_values(ascending=False).round(2)
```
