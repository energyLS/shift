```python
from cmcrameri import cm
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 14, 15)[None, :]
plt.imshow(x, aspect='auto', cmap=cm.batlowS) # or any other colourmap
plt.axis('off')
```

```python
import matplotlib.colors as mcolors

regions = sorted([
    "Central_America", "East_Asia", "East_East_Asia", "Eurasia", "Europe",
    "Far_West_Europe", "Middle_East", "North_America", "North_West_Africa",
    "Oceania", "Pacific_Asia", "South_America", "South_South_America",
    "Subsaharan_Africa", "West_Asia"
])

n = len(regions)
colors_rgba = cm.batlowS(np.linspace(0, 1, n))
hex_colors = [mcolors.to_hex(c).upper() for c in colors_rgba]

for region, hex_color in zip(regions, hex_colors):
    print(f"  {(region + ':').ljust(25)} '{hex_color}'")
```

### okabeito

```python
from cmap import Colormap

okabeito = Colormap('okabeito:okabeito')  # case insensitive
```

```python
x = np.linspace(0, 14, 15)[None, :]
plt.imshow(x, aspect='auto', cmap=okabeito.to_mpl())
plt.axis('off')
```

```python
hex_colors_ok = [mcolors.to_hex(okabeito(i)).upper() for i in np.linspace(0, 1, n)]

for region, hex_color in zip(regions, hex_colors_ok):
    print(f"  {(region + ':').ljust(25)} '{hex_color}'")
```
