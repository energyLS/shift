```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fn = "../../data/un_enerdata_demand_2050_final.csv"
```

```python
df = pd.read_csv(fn, index_col=0)
```

```python
from _helpers import mock_snakemake
snakemake = mock_snakemake(
            "collect_figures",
            scenario="penalty-oc" # penalty-sa, penalty-ea, penalty-nwa, penalty-oc
    )
```

### Prepare data

```python
df["demand"] = df["demand"].div(1e6)
df.unit = "TWh"
df["el_demand"] = df["demand"] * df["el_share"] /100
df.drop(columns=[" note"], inplace=True)

electricity_to_steel = snakemake.config["electricity_steel_ratio"]
df["el_demand in Mt_steel"] = df["el_demand"] / electricity_to_steel
```

```python
df = df.sort_values(by="el_demand", ascending=False)
```

```python
df
```

### Plot data

#### Electricity demand

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(df.index, height=df["el_demand"])

ax.set_ylabel('Electricity demand 2050 in TWh')
ax.set_title('Electricity demand 2050')
ax.grid(axis='y', alpha=0.4, zorder=0)
plt.xticks(rotation=90)

plt.tight_layout()
plt.savefig(snakemake.input.electricity_demand)
plt.savefig(snakemake.input.electricity_demand_png, dpi=300)
plt.show()
```

#### Electricity demand in steel units

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(df.index, height=df["el_demand in Mt_steel"])

ax.set_ylabel('Electricity demand 2050 in Mt steel')
ax.set_title('Electricity demand 2050 in Mt steel')
ax.grid(axis='y', alpha=0.4, zorder=0)
plt.xticks(rotation=90)

plt.tight_layout()
plt.savefig(snakemake.input.electricity_demand_steel)
plt.savefig(snakemake.input.electricity_demand_steel_png, dpi=300)
plt.show()
```
